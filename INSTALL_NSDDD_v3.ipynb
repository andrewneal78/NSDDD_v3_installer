# Frequently Asked Questions & Troubleshooting

## What if something goes wrong?

### Error: "Something timed out"
- **This is normal** - large files can timeout on slow connections
- **Solution**: Simply re-run the cell that failed
- The installer will **resume from where it left off** (no lost data!)
- Repeat if necessary until download completes

### Error: "Failed to create directory" or "Permission denied"
- The notebook doesn't have permission to write to the chosen location
- **Solution**: Choose a different directory, or run from a different location
- **Try**: Desktop, Documents, or your home directory

### Progress shows "Downloaded X MB..." with no percentage
- This happens when the server doesn't report file size
- **It's working** - just can't calculate percentage
- **Solution**: Just be patient, the download will complete

### The notebook keeps asking me to select things
- **This is expected** - each time you run a cell, it asks for input
- You can just click the same options again, or press the button
- The previous values are remembered if you set them

### How do I know if a cell finished?
- **Look for a checkmark** ✓ in the output
- **Green success message** = cell completed
- **Red error message** = something went wrong (can usually retry)
- If you're not sure, scroll to the bottom of the cell output

### Internet connection drops during download
- **No problem!** The installer detects this and retries automatically
- It will retry **up to 3 times** with delays between attempts
- If all retries fail, just re-run the cell to resume

### I want to cancel/stop the installation
- **Press Ctrl+C** (or Cmd+C on Mac) while a cell is running
- The notebook will stop
- **You can always resume later** - downloaded files are saved
- Just re-run the same cell to continue where you stopped

````python
print('='*70)
print('  Downloading Required Files')
print('='*70 + '\n')

downloads_dir = base_path / 'downloads'
required_files = ['model_files', 'metadata', 'whats_new', 'inclusion_criteria', 'readme', 'citation', 'license']

for file_key in required_files:
    download_spec = DOWNLOADS[file_key]
    filename = download_spec['filename']

    print(f'Downloading {filename}...')
    print(f'Size: {download_spec["size_gb"]:.4f} GB\n')

    try:
        file_info = client.get_file_info(filename)
        download_url = file_info['download_url']
        expected_size = file_info['size_bytes']

        destination = downloads_dir / filename

        # Progress callback
        def progress_callback(downloaded, total):
            if total > 0:
                percent = (downloaded / total) * 100
                mb_downloaded = downloaded / (1024 ** 2)
                mb_total = total / (1024 ** 2)
                print(f'  {percent:.1f}% ({mb_downloaded:.0f} MB / {mb_total:.0f} MB)', end='\r')
            else:
                mb_downloaded = downloaded / (1024 ** 2)
                print(f'  Downloaded {mb_downloaded:.0f} MB...', end='\r')

        # Download with resume
        downloaded_file = download_file(
            url=download_url,
            destination=str(destination),
            resume=True,
            progress_callback=progress_callback,
            expected_size=expected_size
        )

        print(f'  ✓ Download complete: {filename}                    \n')

    except Exception as e:
        print(f'  ✗ Error downloading {filename}: {e}')
        print(f'  Note: You can restart this cell to resume downloading.\n')
        raise

print(f'✓ All required files downloaded\n')
````

````python
print('='*70)
print('  Extracting and Verifying Files')
print('='*70 + '\n')

# Import extraction utility
from utils.extract import extract_with_progress

for file_key in required_files:
    download_spec = DOWNLOADS[file_key]
    filename = download_spec['filename']
    extract_to = base_path / download_spec['extract_to']

    # Handle both .zip files and text files
    if filename.endswith('.zip'):
        zip_path = downloads_dir / filename

        print(f'Extracting {filename}...')

        try:
            # Determine target_folder for path stripping
            # model_files.zip contains 'model' folder, metadata.zip contains 'metadata' folder
            target_folder = None
            if file_key == 'model_files':
                target_folder = 'model'
            elif file_key == 'metadata':
                target_folder = 'metadata'

            # Extract with progress and path stripping
            result = extract_with_progress(
                zip_path=zip_path,
                extract_to=extract_to,
                target_folder=target_folder,
                verbose=True
            )

            print(f'  ✓ Extraction complete\n')

        except Exception as e:
            print(f'  ✗ Error extracting {filename}: {e}')
            raise
    else:
        # Copy non-zip files directly
        src = downloads_dir / filename
        dst = extract_to / filename
        
        print(f'Copying {filename}...')
        
        try:
            import shutil
            shutil.copy2(str(src), str(dst))
            print(f'  ✓ Copy complete\n')
        except Exception as e:
            print(f'  ✗ Error copying {filename}: {e}')
            raise

print(f'✓ All files extracted\n')

# Verify model files with checksums
print('Verifying model file integrity...\n')

checksum_file = base_path / 'model' / 'checksums.txt'
if checksum_file.exists():
    try:
        checksums = load_checksums_file(checksum_file)
        all_valid, results = verify_directory(
            base_path / 'model',
            checksums,
            verbose=True
        )

        if all_valid:
            print(f'\n✓ All files verified successfully\n')
        else:
            print(f'\n✗ Verification failed!')
            print(f'  Mismatches: {len(results["mismatches"])}')
            print(f'  Missing: {len(results["missing"])}')
            raise RuntimeError('File verification failed')

    except Exception as e:
        print(f'  ✗ Error during verification: {e}')
        raise
else:
    print('⚠ checksums.txt not found – skipping verification')
````

````python
print('='*70)
print('  Verification Tests')
print('='*70 + '\n')

print('Loading model files...\n')
print('(This takes 90–120 seconds for the 11GB embedding file)\n')

model_dir = base_path / 'model'

try:
    # Load segment encodings (largest file)
    print('Loading segment_encodings.json (11GB)...')
    start = time.time()

    with open(model_dir / 'segment_encodings.json', 'r') as f:
        segment_encodings = json.load(f)
        segment_encodings = np.array(segment_encodings)

    elapsed = time.time() - start
    print(f'  ✓ Loaded: {segment_encodings.shape[0]:,} embeddings ({segment_encodings.shape[1]} dimensions) in {elapsed:.1f}s')

    # Load segments
    print('\nLoading segments_dict.json...')
    with open(model_dir / 'segments_dict.json', 'r') as f:
        segments_dict = json.load(f)
    print(f'  ✓ Loaded: {len(segments_dict):,} segments')

    # Load encoded segments list
    print('Loading encoded_segments.json...')
    with open(model_dir / 'encoded_segments.json', 'r') as f:
        encoded_segments = json.load(f)
    print(f'  ✓ Loaded: {len(encoded_segments):,} segment IDs')

    # Load documents
    print('Loading documents_dict.json...')
    with open(model_dir / 'documents_dict.json', 'r') as f:
        documents_dict = json.load(f)
    print(f'  ✓ Loaded: {len(documents_dict):,} documents')

    # Load countries
    print('Loading countries_dict.json...')
    with open(model_dir / 'countries_dict.json', 'r') as f:
        countries_dict = json.load(f)
    print(f'  ✓ Loaded: {len(countries_dict):,} countries')

    print(f'\n✓ All model files loaded successfully\n')

except Exception as e:
    print(f'✗ Error loading model files: {e}')
    raise
````

````python
print('Testing semantic search...\n')

# Test query
test_query = 'Cyber threats to critical infrastructure'

print(f'Test query: "{test_query}"\n')

try:
    # Encode query
    print('Encoding query...')
    query_embedding = model.encode([test_query])[0]
    print(f'  ✓ Query encoded to {len(query_embedding)} dimensions\n')

    # Compute similarities
    print('Computing similarities against all segments...')
    similarities = []
    for encoding in segment_encodings:
        from scipy.spatial.distance import cosine
        sim = 1 - cosine(query_embedding, encoding)
        similarities.append(sim)

    similarities = np.array(similarities)
    print(f'  ✓ Computed {len(similarities):,} similarity scores\n')

    # Get top results
    top_indices = np.argsort(similarities)[::-1][:10]

    print('Top 10 results:')
    print(f'{"Rank":<5} {"Segment":<15} {"Similarity":<12} {"Preview"}')
    print('─' * 70)

    for rank, idx in enumerate(top_indices, 1):
        segment_id = encoded_segments[idx]
        similarity = similarities[idx]

        # Get segment text
        if segment_id in segments_dict:
            text = segments_dict[segment_id][:50] + '...'
        else:
            text = '[Text not found]'

        print(f'{rank:<5} {segment_id:<15} {similarity:<12.3f} {text}')

    print(f'\n✓ Semantic search test successful\n')

except Exception as e:
    print(f'✗ Error during semantic search test: {e}')
    raise
````
